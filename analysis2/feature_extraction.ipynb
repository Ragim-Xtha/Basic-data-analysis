```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv("anime.csv")
df.head()
```

```python
df.shape
```

```python
df['Title'][0]
```

```python
def ep_count(a):
    check = False
    ep = ""
    for i in a:
        if i == ")":
            break
        
        if i == "(":
            check = True

        if check  == True:
            ep += i


    return ep
```

```python
df["Num_episode"] = df['Title'].apply(ep_count)
```

```python
df.head()
```

```python
df['Num_episode'] = df['Num_episode'].str.replace('(', '', regex=False)
```
